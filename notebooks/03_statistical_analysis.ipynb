# 03 - Statistical Analysis

## Project: Urban Mobility Demand Intelligence

### Objective
Use statistical methods to investigate relationships and differences in hourly bike rental demand.

### Research questions

1. Is temperature significantly correlated with bike rental demand?
2. Is humidity significantly correlated with bike rental demand?
3. Is average rental demand different between working days and non-working days?
4. Does weather condition significantly influence rental demand?

### Statistical techniques
- Pearson correlation test
- Independent samples t-test
- One-way ANOVA
- Hypothesis testing
- P-values and confidence intervals


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import pearsonr, ttest_ind, f_oneway

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("../data/raw/hour.csv")

df.head()
```

```python
correlation, p_value = pearsonr(df["temp"], df["cnt"])

print("Pearson Correlation:", round(correlation, 3))
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("The correlation is statistically significant.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
print(f"Exact p-value: {p_value:.10e}")
```

```python
correlation, p_value = pearsonr(df["hum"], df["cnt"])

print("Humidity Correlation:", round(correlation, 3))
print(f"P-value: {p_value:.10e}")

if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("Humidity has a statistically significant correlation with demand.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
working_days = df[df["workingday"] == 1]["cnt"]
non_working_days = df[df["workingday"] == 0]["cnt"]

t_stat, p_value = ttest_ind(
    working_days,
    non_working_days,
    equal_var=False
)

print("Working day mean:", round(working_days.mean(), 2))
print("Non-working day mean:", round(non_working_days.mean(), 2))
print("T-statistic:", round(t_stat, 3))
print(f"P-value: {p_value:.10e}")

if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("There is a statistically significant difference.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
weather_groups = [
    df[df["weathersit"] == weather]["cnt"]
    for weather in sorted(df["weathersit"].unique())
]

f_stat, p_value = f_oneway(*weather_groups)

print("ANOVA F-statistic:", round(f_stat, 3))
print(f"P-value: {p_value:.10e}")

if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("At least one weather group has a different mean demand.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
print("ANOVA F-statistic:", round(f_stat, 3))
print(f"P-value: {p_value:.10e}")
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
```

```python
tukey_results = pairwise_tukeyhsd(
    endog=df["cnt"],
    groups=df["weathersit"],
    alpha=0.05
)

print(tukey_results)
```

```python

## Statistical Analysis - Key Findings

### 1. Temperature Correlation
- Pearson correlation: 0.405
- Statistically significant at the 5% level.
- Temperature has a moderate positive relationship with rental demand.

### 2. Humidity Correlation
- Pearson correlation: -0.323
- Statistically significant at the 5% level.
- Humidity has a moderate negative relationship with rental demand.

### 3. Working Day Comparison
- Working day mean: 193.21
- Non-working day mean: 181.41
- Welch's independent samples t-test: t = 4.095
- p = 4.249 × 10⁻⁵
- The difference is statistically significant.

### 4. Weather Condition Comparison
- ANOVA F-statistic: 127.174
- p = 1.735 × 10⁻⁸¹
- At least one weather category has a different mean rental demand.

### 5. Tukey HSD Post-hoc Analysis
- Significant differences: Weather categories 1 vs 2, 1 vs 3, and 2 vs 3.
- Other pairwise comparisons were not statistically significant.

## Conclusion

The statistical tests provide evidence that temperature, humidity, working-day status and weather categories are associated with differences in bike rental demand.

These findings will guide the next stage of predictive modelling.
```
