## 01 - Data Audit

### Objective
Understand the structure, quality, and characteristics of the UCI Bike Sharing dataset before performing exploratory data analysis and machine learning.

### Dataset
UCI Bike Sharing Dataset (2011–2012)

### Target Variable
`cnt` - Total number of bike rentals per hour.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("data/raw/hour.csv")

print("Dataset loaded successfully!")
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
df.describe().T
```

```python
missing_values = df.isnull().sum()

print("Missing values per column:")
print(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Unique values per column:")
print(df.nunique())
```

```python
print("Minimum rentals:", df["cnt"].min())
print("Maximum rentals:", df["cnt"].max())
print("Average rentals:", round(df["cnt"].mean(), 2))
print("Median rentals:", df["cnt"].median())
```

```python
df["dteday"] = pd.to_datetime(df["dteday"])

print("Start date:", df["dteday"].min())
print("End date:", df["dteday"].max())
```
